```python
# baseline и preprocessing

from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score, davies_bouldin_score

# preprocessing 1
# посмотрев на гистограммы HUGE distribution для Price, Center/Metro distance, Review_number можно посчитать как выбросы точки с Price > 30000 и Review_number > 1000. Удалим их
df = df.loc[(df.Price < 30000) & (df.Review_number < 1000)]

city_unique_labels = {city: idx for idx, city in enumerate(df['City'].unique())} # кластеры = группы по городам
labels = df['City'].map(city_unique_labels)

# preprocessing 2
print(df.isnull().values.any()) # False => отсутствующих значений нет в датасете
df = pd.get_dummies(df).astype(int)
scaler = StandardScaler()
df = scaler.fit_transform(df)


silhouette = silhouette_score(df, labels)
dbs = davies_bouldin_score(df, labels)

print("silhouette = ", silhouette)
print("davies_bouldin_score =", dbs)
```

```python
# DBSCAN, подбор гиперпараметров

from sklearn.cluster import DBSCAN
import numpy as np

best_silhouette = -float('inf')
best_params = {}

eps = 0.5
while eps < 10:
    for min_samples in (2, 30, 2):
        dbscan_model = DBSCAN(eps=eps, min_samples=min_samples)
        labels = dbscan_model.fit_predict(df)

        clusters_number = np.max(labels) + 1

        if clusters_number <= 1:
            continue
        noises = np.count_nonzero(labels == -1)

        silhouette = silhouette_score(df, labels)
        dbs = davies_bouldin_score(df, labels)

        if silhouette > best_silhouette:
            best_silhouette = silhouette
            best_params = {
                'eps': eps,
                'min_samples': min_samples
            }

    eps+=0.5

print("Best params: ", best_params)
```

```python
dbscan_best_model = DBSCAN(eps=best_params['eps'], min_samples=best_params['min_samples'])
labels = dbscan_best_model.fit_predict(df)

clusters_number = np.max(labels) + 1
noises = np.count_nonzero(labels == -1)

silhouette = silhouette_score(df, labels)
dbs = davies_bouldin_score(df, labels)

print("clusters number = ", clusters_number)
print(f"noises number = {noises} from {df.shape[0]} general number")
print("silhouette = ", silhouette)
print("davies_bouldin_score =", dbs)
```

```python
# собственная реализация DBSCAN

from sklearn.neighbors import KDTree

class Point:
    def __init__(
    self,
    content,
    index,
    label
    ):
        self.index = index,
        self.label = label,
        self.content = content



class MyDBSCAN:
    def __init__(
    self,
    eps = 2,
    cluster_min_size = 5
    ):
        self.eps = eps,
        self.cluster_min_size = cluster_min_size

        self.labels = None


    def expand_cluster(self, cluster_id, point, X, tree): # по сути BFS
        queue = []
        queue.append(point)
        while len(queue) != 0:
            point = queue.pop(0)
            if self.labels[point.index] != -1 and self.labels[point.index] != -2:
                continue
            self.labels[point.index] = cluster_id
            point_neighbours = self.get_neighbours(point, X, tree)
            for neighbour in point_neighbours:
                queue.append(neighbour)

    def get_neighbours(self, point, X, tree):
        neighbours_indices = tree.query_radius([point.content], r=self.eps)[0]
        neighbours = [Point(X[i], i, self.labels[i]) for i in neighbours_indices if i != point.index]

        return neighbours

    def fit(self, X):
        self.labels = np.full(X.shape[0], -1)
        tree = KDTree(X)
        cluster_id = -1
        for i in range(X.shape[0]):
            if self.labels[i] != -1: # the point is already visited
                continue

            point = Point(X[i], i, self.labels[i])
            point_neighbours = self.get_neighbours(point, X, tree)
            if len(point_neighbours) < self.cluster_min_size:
                self.labels[i] = -2 # potential noise
                continue
            cluster_id += 1
            self.expand_cluster(cluster_id, point, X, tree)

        return self.labels
```

Подбор гиперпараметров для собственной модели
При небольшом eps алгоритм работает быстро, но если eps будет около 6.5, то время работы существенно возрастет (т.к. будет получаться (как мы видели на best_dbscan_model) небольшое кол-во кластеров (2), но которые будут очень большими, и мой bfs их будет очень долго обходить, хотя модель будет работать по-прежнему правильно), а если eps перебирать еще и в таком же цикле, как для гиперпараметров моели из sklearn, то результатов придется ждать еще дольше. Поэтому я сделала перебор гиперпараметров не для всего датасета, а для его части.

```python
df = pd.read_csv('hotels_dataset.csv')
df = df.loc[(df.Price < 30000) & (df.Review_number < 1000)]
df = df.sample(frac=1).reset_index(drop=True)
df = pd.get_dummies(df).astype(int)
scaler = StandardScaler()
df = scaler.fit_transform(df)

df_2 = df[:1000]

best_silhouette = -float('inf')
best_params = {}

eps_variants = [1, 3, 6.5]
cluster_min_size_variants = [2, 4, 6]
for eps in eps_variants:
    for cluster_min_size in cluster_min_size_variants:
        my_dbscan_model = MyDBSCAN(eps=eps, cluster_min_size=cluster_min_size)
        labels = my_dbscan_model.fit(df_2)

        clusters_number = np.max(labels) + 1

        if clusters_number <= 1:
            continue
        noises = np.count_nonzero(labels == -1)

        silhouette = silhouette_score(df_2, labels)
        dbs = davies_bouldin_score(df_2, labels)


        if silhouette > best_silhouette:
            best_silhouette = silhouette
            best_params = {
                'eps': eps,
                'cluster_min_size': cluster_min_size
            }

print("Best params: ", best_params)
```

```python
my_dbscan_best_model = MyDBSCAN(eps=best_params['eps'], cluster_min_size=best_params['cluster_min_size'])
labels = my_dbscan_best_model.fit(df_2)

clusters_number = np.max(labels) + 1
noises = np.count_nonzero(labels == -2)

silhouette = silhouette_score(df_2, labels)
dbs = davies_bouldin_score(df_2, labels)

print("clusters number = ", clusters_number)
print(f"noises number = {noises} from {df_2.shape[0]} general number")
print("silhouette = ", silhouette)
print("davies_bouldin_score =", dbs)
```

```python
eps_variants = [1, 3, 6.5]
min_samples_variants = [2, 4, 6]
for eps in eps_variants:
    for min_samples in min_samples_variants:
        dbscan_model = DBSCAN(eps=eps, min_samples=min_samples)
        labels = dbscan_model.fit_predict(df_2)

        clusters_number = np.max(labels) + 1

        if clusters_number <= 1:
            continue
        noises = np.count_nonzero(labels == -1)

        silhouette = silhouette_score(df_2, labels)
        dbs = davies_bouldin_score(df_2, labels)


        if silhouette > best_silhouette:
            best_silhouette = silhouette
            best_params = {
                'eps': eps,
                'min_samples': min_samples
            }

print("Best params: ", best_params)
```
